```python
import math
raio = float(input('Informe o raio da esfera:'))
diametro = 2 * raio
circunferencia = diametro * math.pi
areaSuperf = 4 * math.pi * math.pow(raio , 2)
volume = 4/3 * math.pi * math.pow(raio , 3)
print(f'Diâmetro da esfera:{diametro:.2f}')
print(f'Circunferência da esfera:{circunferencia:.2f}')
print(f'Area da superficie  da esfera:{areaSuperf:.2f}')
print(f'Volume da esfera:{volume:.2f}')
```

```python
#Interando com cifras

cifra = ''

mensagem = 'ANALISE E DESENVOLVIMENTO DE SISTEMAS'
for letra in mensagem:

  if letra == 'a' or letra == 'A':
    cifra += '4'
  elif letra == 'i' or letra == 'I':
    cifra += '1'
  elif letra == 'e' or letra == 'E':
    cifra += '3'
  elif letra == 's' or letra == 'S':
    cifra += '5'
  elif letra == 't' or letra == 'T':
    cifra += '7'
  elif letra == 'g' or letra == 'G':
    cifra += '6'
  elif letra == 'x' or letra == 'X':
    cifra += '#'
  elif letra == 'b' or letra == 'B':
    cifra += '8'
  elif letra == 'o' or letra == 'O':
    cifra += '0'
  else:
    cifra += letra

print(cifra)
```
